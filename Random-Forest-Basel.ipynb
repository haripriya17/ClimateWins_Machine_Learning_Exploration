#### 2. Data Wrangling

Filter the dataset so it only includes data from the Basel station.

```python
# Create a list of the columns containing "Basel" in their names

basel_list = list(x for x in X.columns if x.find('BASEL') >=0)
basel_list
```

```python
# Create a dataframe with those columns

df_basel = X[basel_list]
df_basel
```

```python
# Reduce answers dataset to Heathrow's answers only

answers_basel = answers['BASEL_pleasant_weather']
answers_basel
```

```python
df_basel.shape
```

```python
answers_basel.shape
```

#### 3. Reshaping for modeling

```python
X2 = df_basel
```

```python
y2 = answers_basel
```

```python
# Turn X2 and y2 from df to arrays

X = np.array(X2)
y = np.array(y2)
```

```python
X.shape
```

```python
y.shape
```

#### 4. Data Split

```python
# Split data into train and test sets

X_train, X_test, y_train, y_test = train_test_split(X,y,random_state = 42)
```

```python
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
X_train
```

```python
y_test
```

#### 5. Random Forest Model

```python
# Create a RF classifier
clf = RandomForestClassifier(n_estimators = 100)#, max_depth=5)  
  
# Training the model on the training dataset
# fit function is used to train the model using the training sets as parameters
clf.fit(X_train, y_train)
```

```python
# Perform predictions on the test dataset
y_pred = clf.predict(X_test)
  
# Use metrics module for accuracy calculation
print("Model Accuracy: ", metrics.accuracy_score(y_test, y_pred))
```

```python
# Class-names = {0:'Unpleasant Weather', 1:'Pleasant Weather'}

fig = plt.figure(figsize=(80,40))
plot_tree(clf.estimators_[6], fontsize = 20, filled=True);
```

```python
fig.savefig(os.path.join(path, '04 Analysis','Visualizations', 'random_forest_Basel.png'),bbox_inches='tight')
```

#### 6. Uncovering Feature Importances

```python
# Retrieve feature importances from the trained model

newarray = clf.feature_importances_
print(clf.feature_importances_.shape)
newarray
```

```python
# Create a list of weather features

wx_list = [feature.replace('BASEL_', '') for feature in basel_list]
wx_list
```

```python
important = pd.Series(newarray, index = wx_list)
important
```

```python
# Plot the results

%matplotlib inline

plt.style.use('fivethirtyeight')
# list of x locations for plotting
x_values = list(range(len(newarray)))
print(wx_list)

plt.bar(x_values, newarray, orientation = 'vertical')
plt.xticks(x_values, wx_list, rotation='vertical')
plt.ylabel('Importance'); plt.xlabel('Feature'); plt.title('Features Importances for Basel (all years)');

plt.savefig(os.path.join(path, '04 Analysis','Visualizations', 'Basel_feature_importances.png'), bbox_inches='tight')
plt.show()
```
